```python
#Real data
Oscar_movies = [
    "The dark knight",
    "The Hurt Locker",
    "The King Speech",
    "The Artist",
    "Argp"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

#plot
plt.plot(years, oscar_revenue)
plt.title("Oscar movie revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue")
```

```python
#non oscar movies 
non_oscar_movies = ["Slumbong", "Avatar", "Inception", "Hugo", "Lincoin"]
non_oscar_revenue = [378, 2788, 829, 185, 275]

plt.plot(years, oscar_revenue , color="red", marker="<", linestyle="-.",linewidth="5", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")

plt.title("Oscar Movies vs non_oscar movies ")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
#plt.style.use("default")
plt.grid()
plt.show()
```

```python
#bar charts 
movies = [
    "The dark knight",
    "The hurt Locker",
    "The king's speech",
    "The artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 178, 427, 133, 232]

plt.bar(years, revenue, color="green")
plt.title("Revenue in each year")
plt.xlabel("Year")
plt.ylabel("Reveneu")

for i in range(len(years)):
    plt.text(years[i], revenue[i]+20, str(revenue[i]), ha="center")

plt.ylim(0, max(revenue)+100)
```

```python
#multiple plots
cities = ["City A", "City B", "City C", "City D", "City E"]

winter_temp = [5, 2, 10, 8, 7]
winter_humidity = [80, 75, 65, 85, 70]

summer_temp = [25, 30, 28, 27]
summer_humidity = [60, 50, 55, 65, 45]

plt.scatter(winter_temp ,winter_humidity , label="winter")
plt.scatter(summer_temp, summer_humidity , label="summer")

plt.title("Summer vs winter data")
plt.xlabel("Temp(C)")
plt.ylabel("Humidity (%)")
plt.legend()
```

```python
#pie charts 
expences = ["Salaries", "rent", "marketing", "R&D", "Miscelleneous"]
amounts=[500,150,50,80,100]

plt.style.use("default")
#colors=["red","black","blue","pink","yellow"]

explode = [0, 0, 0,0.2, 0]
plt.pie(amounts, labels=expences, autopct="%1.1f%%",
       wedgeprops = {
           "edgecolor":"black",
           "linewidth":1,
           "linestyle":"--"
       },
        explode=explode,
        startangle=180,
        shadow=True
       )
plt.show()
```
