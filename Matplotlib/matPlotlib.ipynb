```python
#Real data
Oscar_movies = [
    "The dark knight",
    "The Hurt Locker",
    "The King Speech",
    "The Artist",
    "Argp"
]
years = [2008, 2009, 2010, 2011, 2012]
oscar_revenue = [1005, 170, 427, 133, 232]

#plot
plt.plot(years, oscar_revenue)
plt.title("Oscar movie revenue in each year")
plt.xlabel("Years")
plt.ylabel("Revenue")
```

```python
#non oscar movies 
non_oscar_movies = ["Slumbong", "Avatar", "Inception", "Hugo", "Lincoin"]
non_oscar_revenue = [378, 2788, 829, 185, 275]

plt.plot(years, oscar_revenue , color="red", marker="<", linestyle="-.",linewidth="5", label="Oscar Movies")
plt.plot(years, non_oscar_revenue, label="Non-Oscar Movies")

plt.title("Oscar Movies vs non_oscar movies ")
plt.xlabel("Year")
plt.ylabel("Revenue")
plt.legend()
#plt.style.use("default")
plt.grid()
plt.show()
```

```python
#bar charts 
movies = [
    "The dark knight",
    "The hurt Locker",
    "The king's speech",
    "The artist",
    "Argo"
]
years = [2008, 2009, 2010, 2011, 2012]
revenue = [1005, 178, 427, 133, 232]

plt.bar(years, revenue, color="green")
plt.title("Revenue in each year")
plt.xlabel("Year")
plt.ylabel("Reveneu")

for i in range(len(years)):
    plt.text(years[i], revenue[i]+20, str(revenue[i]), ha="center")

plt.ylim(0, max(revenue)+100)
```
